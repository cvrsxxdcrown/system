##  О датасете
[Pima Indians Diabetes Database](https://www.kaggle.com/datasets/uciml/pima-indians-diabetes-database/data)

Этот набор данных взят из Национального института диабета, заболеваний органов пищеварения и почек (NIDDK). Целью набора данных является диагностическое предсказание наличия у пациента диабета на основе определенных диагностических измерений, включенных в набор данных. На выбор этих экземпляров из более крупной базы данных было наложено несколько ограничений. В частности, все пациенты здесь — женщины не моложе 21 года, родом из индейцев Пима.

- `Insulin` - 2-часовой сывороточный инсулин (мкЕд/мл);
- `BMI` - Индекс массы тела (вес в кг/(рост в м)^2);
- `DiabetesPedigreeFunction` - функция оценки вероятности диабета на основе семейного анамнеза;
- `Age` - возраст (лет);
- `Outcome` - целевая переменная (0 или 1) 268 из 768 — 1 (больны), остальные — 0 (здоровы).

## Метод PCA

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

target_col = "Outcome" if "Outcome" in df.columns else df.columns[-1]
X = df.drop(columns=[target_col]).copy()
y = df[target_col].copy()

# Часто в diabetes датасете 0 в некоторых столбцах означает "нет данных"
cols_with_zeros_as_missing = ["Glucose","BloodPressure","SkinThickness","Insulin","BMI"]
for col in cols_with_zeros_as_missing:
    if col in X.columns:
        X[col] = X[col].replace(0, np.nan)

# Заполним пропуски медианой
X = X.fillna(X.median(numeric_only=True))

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total explained variance:", pca.explained_variance_ratio_.sum())

plt.figure(figsize=(7,5))
plt.scatter(X_pca[:,0], X_pca[:,1], c=y, alpha=0.7)
plt.title("PCA (2 компоненты)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.grid(True, alpha=0.3)
plt.show()
```

## Метод LDA

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

# Для бинарной классификации LDA даёт максимум 1 компоненту (n_classes - 1)
lda = LinearDiscriminantAnalysis(n_components=1)
X_lda = lda.fit_transform(X_scaled, y)

print("LDA shape:", X_lda.shape)

plt.figure(figsize=(7,4))
plt.scatter(X_lda[:,0], np.zeros_like(X_lda[:,0]), c=y, alpha=0.7)
plt.title("LDA (1 компонент)")
plt.xlabel("LD1")
plt.yticks([])
plt.grid(True, alpha=0.3)
plt.show()
```

## Метод TruncatedSVD

```python
from sklearn.decomposition import TruncatedSVD

svd = TruncatedSVD(n_components=2, random_state=42)
X_svd = svd.fit_transform(X_scaled)

print("Explained variance ratio (approx):", svd.explained_variance_ratio_)
print("Total explained variance (approx):", svd.explained_variance_ratio_.sum())

plt.figure(figsize=(7,5))
plt.scatter(X_svd[:,0], X_svd[:,1], c=y, alpha=0.7)
plt.title("TruncatedSVD (2 компоненты)")
plt.xlabel("SVD1")
plt.ylabel("SVD2")
plt.grid(True, alpha=0.3)
plt.show()
```

## Метод Kernel PCA

```python
# KernelPCA: в следующих ячейках используем разные ядерные функции
```

###  Ядерная функция `linear`

```python
from sklearn.decomposition import KernelPCA

kpca = KernelPCA(n_components=2, kernel="linear", random_state=42)
X_kpca = kpca.fit_transform(X_scaled)

plt.figure(figsize=(7,5))
plt.scatter(X_kpca[:,0], X_kpca[:,1], c=y, alpha=0.7)
plt.title("KernelPCA kernel=linear")
plt.xlabel("KPC1")
plt.ylabel("KPC2")
plt.grid(True, alpha=0.3)
plt.show()
```

###  Ядерная функция `poly`

```python
from sklearn.decomposition import KernelPCA

kpca = KernelPCA(n_components=2, kernel="poly", degree=3, random_state=42)
X_kpca = kpca.fit_transform(X_scaled)

plt.figure(figsize=(7,5))
plt.scatter(X_kpca[:,0], X_kpca[:,1], c=y, alpha=0.7)
plt.title("KernelPCA kernel=poly (degree=3)")
plt.xlabel("KPC1")
plt.ylabel("KPC2")
plt.grid(True, alpha=0.3)
plt.show()
```

###  Ядерная функция `rbf`

```python
from sklearn.decomposition import KernelPCA

kpca = KernelPCA(n_components=2, kernel="rbf", gamma=None, random_state=42)
X_kpca = kpca.fit_transform(X_scaled)

plt.figure(figsize=(7,5))
plt.scatter(X_kpca[:,0], X_kpca[:,1], c=y, alpha=0.7)
plt.title("KernelPCA kernel=rbf")
plt.xlabel("KPC1")
plt.ylabel("KPC2")
plt.grid(True, alpha=0.3)
plt.show()
```

###  Ядерная функция `sigmoid`

```python
from sklearn.decomposition import KernelPCA

kpca = KernelPCA(n_components=2, kernel="sigmoid", random_state=42)
X_kpca = kpca.fit_transform(X_scaled)

plt.figure(figsize=(7,5))
plt.scatter(X_kpca[:,0], X_kpca[:,1], c=y, alpha=0.7)
plt.title("KernelPCA kernel=sigmoid")
plt.xlabel("KPC1")
plt.ylabel("KPC2")
plt.grid(True, alpha=0.3)
plt.show()
```

###  Ядерная функция `cosine`

```python
from sklearn.decomposition import KernelPCA

kpca = KernelPCA(n_components=2, kernel="cosine", random_state=42)
X_kpca = kpca.fit_transform(X_scaled)

plt.figure(figsize=(7,5))
plt.scatter(X_kpca[:,0], X_kpca[:,1], c=y, alpha=0.7)
plt.title("KernelPCA kernel=cosine")
plt.xlabel("KPC1")
plt.ylabel("KPC2")
plt.grid(True, alpha=0.3)
plt.show()
```

###  Ядерная функция `precomputed`

```python
from sklearn.decomposition import KernelPCA
from sklearn.metrics.pairwise import pairwise_kernels

# Считаем матрицу ядра заранее (пример: rbf-ядро)
K = pairwise_kernels(X_scaled, metric="rbf")

kpca = KernelPCA(n_components=2, kernel="precomputed", random_state=42)
X_kpca = kpca.fit_transform(K)

plt.figure(figsize=(7,5))
plt.scatter(X_kpca[:,0], X_kpca[:,1], c=y, alpha=0.7)
plt.title("KernelPCA kernel=precomputed (rbf kernel matrix)")
plt.xlabel("KPC1")
plt.ylabel("KPC2")
plt.grid(True, alpha=0.3)
plt.show()
```
